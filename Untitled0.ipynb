Recordemos que: MCD(a, b) = MCD(b, a mod b)
• Implementa mcd_iter(a, b) usando un ciclo while: calcula r = a % b, luego actualiza a ← b y b ←
r.
• El proceso termina cuando b = 0; el MCD es el último valor de a.
• Calcula e imprime MCD(84, 30) y MCD(252, 105)

```python

def mcd_iter(a,b):
  while b !=0 :
    r=  a % b
    a=b
    b=r
    print("MCD (",a, ",",b,")")
  return a
mcd_iter(84,30)
```
